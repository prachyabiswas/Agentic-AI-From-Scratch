# Vector DB Basics (ChromaDB)

Before wiring up a full RAG chain, I wanted to play with a vector database on its own and
actually *see* semantic search work. [Chroma](https://www.trychroma.com/) is a free,
open-source vector DB that's easy to run in-memory for exactly this kind of experiment.

See [`../../docs/08_rag_theory.md`](../../docs/08_rag_theory.md) for the theory behind what's
happening here.

```python
import chromadb

client = chromadb.Client()  # in-memory client, nothing persisted to disk
```

## Create a collection

A **collection** is roughly the vector-DB equivalent of a table — a named group of
embeddings you can add to and query.

```python
collection = client.create_collection("news")
```

## Add some documents

Chroma handles the embedding step automatically here (using a default embedding model) — we
just hand it raw text and IDs. Deliberately mixing an unrelated topic in (`Apples are high in
fiber...`) to make the semantic search meaningfully distinguish from near-miss keyword
matches.

```python
collection.add(
    ids=["id1", "id2", "id3", "id4"],
    documents=[
        "Apple is leading the smartphone market with iPhone sales up 35%.",
        "Tesla booked a profit of 1 billion dollars in Q2.",
        "Apples are high in fiber, vitamin C, and various antioxidants.",
        "SpaceX secured a NASA contract worth 10 billion dollars.",
    ],
)
```

## Query by meaning, not by keyword

This is the whole point of a vector DB: none of these queries share exact words with the
"right" document, yet semantic similarity finds it anyway.

```python
results = collection.query(
    query_texts=["This is a query related to Elon Musk"],
    n_results=2,
)
print(results["documents"])
```

```python
results = collection.query(
    query_texts=["I love iPad"],
    n_results=2,
)
print(results["documents"])
```

```python
results = collection.query(
    query_texts=["Tips for improving health"],
    n_results=2,
)
print(results["documents"])
```

Notice: the "Elon Musk" query surfaces the Tesla/SpaceX documents, "I love iPad" surfaces
the Apple/iPhone document (not the apple-the-fruit one), and "tips for improving health"
surfaces the fiber/antioxidants document — all without a single shared keyword. That's
semantic search in action.
